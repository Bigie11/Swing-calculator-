1. what is data science?
Data science can be defined as the cllection, analysis, organizing and interpretation of data to proffer better solution to you and make better dicisions.

2. Why should a data analyst inspect data before making conclusions?
A data analyst should inspect data first to check for errors, missing values, duplicates, and inconsistencies. This helps ensure the data is accurate and reliable, leading to correct conclusions and better decisions.

3. Why are missing values important in real datasets?
Missing values are important because they can affect the accuracy of data analysis and produce misleading results if ignored. Identifying and handling them properly helps improve the quality and reliability of the analysis.
